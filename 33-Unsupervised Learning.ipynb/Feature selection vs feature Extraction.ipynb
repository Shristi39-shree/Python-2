#### Why diensionality Reduction??
* Prevent curse of dimensionality
* To improve the performance of the model
* To visualize the data 


* x++ y++, x-- y-- -->Covariance positive
* x-- y++, x++ y-- -->Covariance Negative
* Covariance = 0: No relationship between x and y(Scattered plot)

* Covariance(x,y) = Sum(xi-mean(x))(yi-mean(y))/N-1

* Pearson Correlation = Cov(x,y)/SDx*SDy -->range(-1,1)
* The more value towards +1 the more +ve correlated x and y is .Vice versa
* and near to 0 means no relation

* DATA HOUSING(Example)
1) House size(Independent feature) 2) Fountain size(Independent feature) 3) Price(output feature)
* Graph: fountain size vs Price --> linear(covariance = +ve)
* Grapg: fountain size vs house size --> constant(covariance = 0)

#### Feature Extraction

* Room size ,No of rooms,Price ---> House size and price (2 feature - 1 feature-Dimensionality reduction)
