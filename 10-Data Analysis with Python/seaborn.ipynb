### Data Visualisation With Seaborn

Seaborn is a Python visualisation library based on Matplotlib that provides a high-level interface for drawing attractive and informative statistical graphics.Seaborn helps in creating complex visualizations with just a few lines of code.In this lesson, we will cover the basics of Seaborn,including creating various types of plots and customizing them.

```python
import seaborn as sns
```

```python
### Basic plotting eith Seaborn
tips=sns.load_dataset('tips')
tips
```

```python
#create a scatter plot
import matplotlib.pyplot as plt

sns.scatterplot(x='total_bill',y='tip',data=tips)
plt.title("Scatter Plot of Total Bill vs Tip")
plt.plot()
```

```python
## Line Plot

sns.lineplot(x='size',y='total_bill',data=tips)
plt.title("Line Plot of Total bill by size")
plt.show()
```

```python
## Categorical Plots
## BAr Plot

sns.barplot(x='day',y='total_bill',data=tips)
plt.title('Bar Plot of Total Bill By Day')
plt.show()
```

```python
## Box Plot
sns.boxplot(x='day',y='total_bill',data=tips)
```

```python
## Violin Plot

sns.violinplot(x='day',y='total_bill',data=tips)
```

```python
### Histograms
sns.histplot(tips['total_bill'],bins=10,kde=False)
```

```python
### Histograms
sns.histplot(tips['total_bill'],bins=10,kde=True)
```

```python
## KDE Plot
sns.kdeplot(tips['total_bill'],shade=True)
```

```python
#Pairplot
sns.pairplot(tips) 
```

```python
## Heatmap
corr=tips[['total_bill','tip','size']].corr()
corr
```

```python
sns.heatmap(corr,annot=True,cmap='coolwarm')
```

```python
import pandas as pd
sales_df=pd.read_csv('sales_data.csv', encoding='cp1252')
sales_df.head()
```

```python
## Plot total sales by Dealsize
plt.figure(figsize=(10,6))
sns.barplot(x='DEALSIZE',y="SALES",data=sales_df,estimator=sum)
plt.title('Total Sales by Product')
plt.xlabel('Product')
plt.ylabel('Total Sales')
plt.show()
```

```python
## Plot total sales by Dealsize
plt.figure(figsize=(10,6))
sns.barplot(x='DEALSIZE',y="SALES",data=sales_df,estimator=sum)
plt.title('Total Sales by Dealsize')
plt.xlabel('Dealsize')
plt.ylabel('Total Sales')
plt.show()
```
