## Data Manipulation and Analysis with Pandas 

Data manipulation and analysis are key tasks in any data science or data analysis project.Pandas provides a wide range of functions for data manipulation and analysis ,making it easier to clean,transform, and extract insights from data.In this lesson,we will cover various data manipulation and analysis techniques using Pandas.

```python
import pandas as pd
```

```python
df = pd.read_csv('sales_data.csv', encoding='latin1')

## fetch the first 5 rows
df.head(5)
```

```python
## Handling Missing values
df[df.isnull()]
```

```python
df_filled=df.isnull().sum()
```

```python
df.fillna(0)
```

```python
### filling missing values with the mean of the column 
df['PRICEEACH_fillNA']=df['PRICEEACH'].fillna(df['PRICEEACH'].mean())
df
```

```python
df.dtypes
```

```python
## Renaming Column
df=df.rename(columns={'SALES':'Product Sales'})
df.head()
```

```python
## change datatypes
# df['Value_new'] = df['ORDERNUMBER'].astype(float)

df['Value_new'] = df['ORDERNUMBER'].fillna(df['ORDERNUMBER'].mean()).astype(int)
df.head()
```

```python
df['New Value'] = df['Value_new'].apply(lambda x:x*2)
df.head()
```

```python
## Data Aggregating and Grouping
df.head()
```

```python
grouped_mean=df.groupby('COUNTRY')['Value_new'].mean()
print(grouped_mean)
```

```python
grouped_sum=df.groupby(['COUNTRY','DEALSIZE'])['Value_new'].sum()
print(grouped_sum)
```

```python
## Agrregate multiple functions
grouped_agg=df.groupby('COUNTRY')['Value_new'].agg(['mean','sum','count'])
grouped_agg
```

```python
## Merging and joining dataframes
# Create sample Dataframes
df1=pd.DataFrame({'Key': ['A','B','C'], 'Value1':[1,2,3]})
df2=pd.DataFrame({'Key': ['A','B','D'], 'Value2':[4,5,6]})
```

```python
df1
```

```python
df2
```

```python
# Merge Dataframe on the 'Key Columns'
pd.merge(df1,df2,on="Key",how="inner")
```

```python
pd.merge(df1,df2,on="Key",how="outer")
```

```python
pd.merge(df1,df2,on="Key",how="left")
```

```python
pd.merge(df1,df2,on="Key",how="right")
```
