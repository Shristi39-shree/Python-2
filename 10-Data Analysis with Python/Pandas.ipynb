## Pandas-DataFrame And Series

Pandas is a powerful data manipulation library in Python ,widely used for data analysis and data cleaning.It provides two primary data structures : Series and DataFrame.A Series is a one-dimensional array-like object,while a DataFrame is a two-dimensional,size-mutable, and potentially heterogenous tabular data structure with labelled axes(rows and columns).

```python
## Series
## A Pandas Series is a one-dimensional array-like object that can hold any data type.It is similar to a column in the table.

import pandas as pd
data = [1,2,3,4,5]
series=pd.Series(data)
print("Series \n",series)
print(type(series))
```

```python
 # Create a Series from dictionary
data = {'a':1,'b':2,'c':3}
series_dict=pd.Series(data)
print(series_dict)
```

```python
data = [10,30,40]
index = ['a','b','c']
pd.Series(data,index=index)
```

```python
# Dataframe
# Create a Dataframe from a dictionary of list

data = {
  'Name':['Krish','John','Jack'],
  'Age':[30,40,20],
  'City':['Banglore','Benaras','New York']
}
df=pd.DataFrame(data)
print(df)
print(type(df))
```

```python
# Create a Data frame from a List of Dictionaries

data = [
  {'Name':'Krish','Age':30,'City':'Bbsr'},
   {'Name':'Shree','Age':20,'City':'Patna'},
    {'Name':'Sant','Age':2,'City':'Patna'}  
]
df = pd.DataFrame(data)
print(df)
print(type(df))
```
