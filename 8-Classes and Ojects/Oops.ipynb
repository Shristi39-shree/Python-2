### Classes and Objects

Object-Oriented Programming(OOP) is a programming paradigm that uses "objects" to design applications and computer programs.OOP allows for modeling real-world scenerios using classes and objects.This lesson covers the basics of creating classes and objects,including instance variables and methods.

```python
### A class is a blue print for creating objects.Attributs,methods

class Car:
   pass
audi = Car()
bmw = Car()

print(type(audi))
```

```python
### Instance Variable and Methods
class Dog:
    ## constructor
    def __init__(self,name,age):
        self.name =name
        self.age = age
## create object 
dog1 = Dog("Buddy",3)
print(dog1) 
print(dog1.name)
print(dog1.age)       
```

```python
dog2 = Dog("Luccy",4)
print(dog2.name)
print(dog2.age)
```

```python
## define a class with instance methods 
class Dog:
  def __init__(self,name,age):
    self.name = name
    self.age = age

  def bark(self):                   ### Method 
    print(f"{self.name} says woof")

dog1=Dog("Buddy",3)
dog1.bark()
dog2=Dog("Lucy",6)    
dog2.bark()  
```

```python
### modeling a Bank account 


## Define a class for bank account 

class BankAccount:
  def __init__(self,owner,balance = 0):
    self.owner = owner
    self.balance = balance

  def deposit(self,amount):
     self.balance+=amount
     print(f"{amount} is deposited .New balance is {self.balance}")

  def withdraw(self,amount):
     if amount>self.balance:
        print("Insufficient balance")
     else :
        self  .balance -= amount
        print(f"{amount} is withdrawn.New balance is {self.balance}")

  def get_balance(self):
     return self.balance


## create an account 

account = BankAccount("Shristi",20000) 
print(account.balance)       
```

```python
## Call instance methods
account.deposit(1000)
```

```python
account.withdraw(200)
```

```python
account.withdraw(90000)
```

```python
print(account.get_balance())
```
