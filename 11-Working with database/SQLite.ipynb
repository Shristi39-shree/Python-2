### SQL And SQLite

SQl(Structured Query Language) is a standard language for managing and manipulating relational database.
SQLite is a self-contained,severless, and zero-configuration database engine that is widely used for embedded database system.In this lesson,we will cover the basics of SQL and SQLite,including creating database,tables, and performing various SQL operations.

```python
import sqlite3
```

```python
## Connect to an SQLite database
connection=sqlite3.connect('example.db')
connection
```

```python
cursor=connection.cursor()  ## To iterate table and rows
```

```python
## Create a Table
cursor.execute('''
Create Table If Not Exists employees(
    id Integer Primary Key,
    name Text Not Null,
    age Integer,
    department text
    )
''')

##Commit the changes
connection.commit()
```

```python
cursor.execute('''
Select * from employees
  
''')
```

```python
## INsert the data in sqlite table
cursor.execute('''
Insert Into employees(name,age,department)
               values('Priyu',20,'Business Man')
               ''')  

cursor.execute('''
Insert Into employees(name,age,department)
               values('Shree',20,'Engineer')
               ''') 

cursor.execute('''
Insert Into employees(name,age,department)
               values('Shristi',20,'IAS Officer')
               ''')                

##commit the changes 
connection.commit()     
```

```python
## Query the data from the table
cursor.execute('Select * from employees')
rows=cursor.fetchall()

## print the queried data

for row in rows:
  print(row)
```

```python
## Update the data in the table
cursor.execute('''
UPDATE employees
Set age =50
where name='Priyu'                             
''' )

connection.commit()
```

```python
## Query the data from the table
cursor.execute('Select * from employees')
rows=cursor.fetchall()

## print the queried data

for row in rows:
  print(row)
  
```

```python
## Delete the data from the table
cursor.execute('''
Delete from employees
               where name='Priyu'
''')
```

```python
## Query the data from the table
cursor.execute('Select * from employees')
rows=cursor.fetchall()

## print the queried data

for row in rows:
  print(row)
  
```

```python
## Working with Sales data
connection = sqlite3.connect('sales_data.db')
cursor = connection.cursor()

# Create a table for sales data 
cursor.execute('''   
CREATE  TABLE IF NOT EXISTS sales (
    id INTEGER PRIMARY KEY,
    product_number INTEGER NOT NULL,
    product TEXT NOT NULL,
    sales INTEGER,       
    region TEXT  
)                                       
''')

# Insert data into the table 
sales_data=[
  ('200034','Product1',100,'North'),
  ('200035','Product2',200,'South'),
  ('200036','Product3',300,'East'),
  ('200037','Product4',400,'West'),
  ('200038','Product5',500,'North-east')
]

## Bulk Insertion
cursor.executemany('''
Insert into sales(product_number,product,sales,region)
                   values(?,?,?,?)
''',sales_data)

connection.commit()
```

```python
## Query the data from the table
cursor.execute('Select * from sales')
rows=cursor.fetchall()

## print the queried data

for row in rows:
  print(row)
```

```python
##CLose the connection
connection.close()
```
