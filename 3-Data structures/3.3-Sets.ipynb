### Sets

##### Sets are a build -in data type in Python used to store collections f unique items.They are unordered,meaning thet the elements do not follow a specific order and they do not allow duplicate elements.Sets are useful for membership tests,eliminating duplicate enteries, and performing mathematical set operations like union,intersection,difference ,and symmetric difference.

```python
## Create a set
my_set={1,2,3,4,5}
print(my_set)
print(type(my_set))
```

```python
my_empty_set=set()
print(type(my_empty_set))
```

```python
my_empty_set=set([1,2,3,4,5])
print(my_empty_set)
```

```python
## Basics Sets Operation
## Adding and Removing ELements
my_set.add(7)
print(my_set)
my_set.add(7)
print(my_set)
```

```python
## Remove the elements from a set 
my_set.remove(3)
print(my_set)
```

```python
my_set.remove(10)
print(my_set)
```

```python
my_set.discard(11)
print(my_set)
```

```python
## pop method
removed_element=my_set.pop()
print(removed_element)
print(my_set)
```

```python
## clear all the elements

my_set.clear()
print(my_set)
```

```python
## Set Membership test
my_set = {1,2,3,4,5}
print(2 in my_set)
print(10 in my_set)
```

```python
## mathematical Operations
set1 = {1,2,3,4,5,6}
set2 = {7,8,9,3,4,8,10}

### Union
union_set = set1.union(set2)
print(union_set)

### Intersection
intersection_set = set1.intersection(set2)
print(intersection_set)

set1.intersection_update(set2)
print(set1)
```

```python
set1 = {1,2,3,4,5,6}
set2 = {7,8,9,3,4,8,10}

## Difference
print(set1.difference (set2))
print(set2.difference(set1))
```

```python
## symmetric Difference
set1 = {1,2,3,4,5,6}
set2 = {7,8,9,3,4,8,10}

set1.symmetric_difference(set2)
```

```python
## sets Methods
set1={12,23,45}
set2 = {14.13,45}

## is subset
print(set1.issubset(set2))

## is superset
print(set1.issuperset(set2))
```

```python
set1={1,2,3}
set2 = {1,2,3,4,5,6,7}

## is subset
print(set1.issubset(set2))

## is superset
print(set2.issuperset(set1))
```

```python
lst=[1,2,3,4,5,5,6]
set(lst)
```

```python
### Counting Unique words in text

text="In this tutorial we are discussing about sets"
words =text.split()

 ## convert list of words to set to get unique words

unique_words=set(words)
print(unique_words)
print(len(unique_words))
```
