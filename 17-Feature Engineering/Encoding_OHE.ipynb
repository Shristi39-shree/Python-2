### Nominal/OHE Encoding
One hot encoding ,also known as nominal encoding is a technique used to categorical data as numerical data,which is more suiable for ml algorithms.In this technique,each category is represented as a binary vector where each bit corresponds to a unique category.For example,if we have a categorical variable "color" with the possible values(red,green,blue),we can represent it in one hot encoding as follows:

1.Red:[1,0,0]

2.Green:[0,1,0]

3.Blue:[0,0,1]

```python
import pandas as pd
from sklearn.preprocessing  import OneHotEncoder
```

```python
### Create a simple Dataframe
df = pd.DataFrame({
  'color':['red','blue','green','green','red','blue']
})
```

```python
df.head()
```

```python
## Create an instance of OneHotEncoder
encoder= OneHotEncoder()
```

```python
## perform fit and transform
encoded=encoder.fit_transform(df[['color']]).toarray()
```

```python
import pandas as pd
encoder_df=pd.DataFrame(encoded,columns=encoder.get_feature_names_out())
```

```python
encoder_df
```

```python
encoder.transform([['blue']]).toarray()
```

```python
pd.concat([df,encoder_df],axis=1)
```
